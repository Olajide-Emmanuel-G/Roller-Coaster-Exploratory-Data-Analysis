```python
# STEP 4: FEATURE RELATIONSHIPS
```

```python
# Scatter plot for Speed_mph and Height_ft column
df.plot(kind="scatter", x="Speed_mph", y="Height_ft", title="Coaster Speed  Vs Height")

# Adding the show function to make the plot more cleaner 
plt.show()
```

```python
# Making an advanced plot analysis using seaborn
sns.scatterplot( x="Speed_mph", 
                 y="Height_ft", 
                 data=df)
plt.show()
```

```python
# Advanced plot based on the year the roller coaster was introduced
sns.scatterplot( x="Speed_mph", 
                 y="Height_ft", 
                 hue="Year_Introduced",
                 data=df)
plt.show()
```

```python
# Comparing more than two features against each other
sns.pairplot(df, vars=["Year_Introduced", "Speed_mph", "Height_ft", "Inversions", "Gforce"],
            hue="Type_Main")
plt.show()
```

```python
 # Comparing features together through correlation
df_corr = df[
     ["Year_Introduced", "Speed_mph", "Height_ft", "Inversions", "Gforce"]
     ].dropna().corr()
df_corr
```

```python
# Comparing correlation using seaborn heatmap
sns.heatmap(df_corr, annot=True)
plt.show()
```

```python
# What are the location with the fastest roller coasters(minimm of 10)?
df["Location"].value_counts()
```

```python
# Quering where location is not equal to others to filter out the rest of the location values
ax = df.query("Location != 'Other'")\
    .groupby("Location")["Speed_mph"]\
    .agg(["mean", "count"])\
    .query("count >= 10")\
    .sort_values("mean")["mean"]\
    .plot(kind="barh", figsize=(12, 5), title="Average Coaster Speed by Location")
ax.set_xlabel("Average Coaster Speed")
plt.show()
```
